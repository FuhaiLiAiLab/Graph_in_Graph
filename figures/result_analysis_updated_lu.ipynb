```python
import numpy as np 
import matplotlib.pyplot as plt
from matplotlib.collections import PatchCollection
from matplotlib.colors import TwoSlopeNorm
import seaborn as sns
import pandas as pd

filtered_genes = t2ds_sort_norm_refilter_node_weight_df['gene_node_name'].tolist()

tvsno_pvalue_files = [
    './pvalues_output/TvsNO/TvsNO_core_promoter_pvalues.csv',
    './pvalues_output/TvsNO/TvsNO_distal_promoter_pvalues.csv',
    './pvalues_output/TvsNO/TvsNO_downstream_pvalues.csv',
    './pvalues_output/TvsNO/TvsNO_proximal_promoter_pvalues.csv',
    './pvalues_output/TvsNO/TvsNO_tran_v1_pvalues.csv',
    './pvalues_output/TvsNO/TvsNO_upstream_pvalues.csv'
]

def combine_pvalue_files(pvalue_files, filtered_genes):
    combined_pvalues = []
    for pfile in pvalue_files:
        pvalues_df = pd.read_csv(pfile)
        filtered_pvalues = pvalues_df[pvalues_df['gene'].isin(filtered_genes)]['adj_p_value'].tolist()  # ← 改这里
        combined_pvalues.append(filtered_pvalues)
    return combined_pvalues

tvsno_pvalues = combine_pvalue_files(tvsno_pvalue_files, filtered_genes)
pval_list_of_lists = tvsno_pvalues 

xlabels = ['TvsNO_core_promoter', 'TvsNO_distal_promoter', 'TvsNO_downstream', 'TvsNO_proximal_promoter',
           'TvsNO_tran_v1', 'TvsNO_upstream']

ylabels = filtered_genes

xlabels_num_list = list(np.arange(0, len(xlabels))) * len(ylabels)
ylabels_num_list = sum([[i] * len(xlabels) for i in range(len(ylabels))], [])

s = np.concatenate(pval_list_of_lists)
c = np.concatenate(pval_list_of_lists)

fig, ax = plt.subplots(figsize=(30, 10))
ax.set_xlim(-0.5, len(xlabels) - 0.5)
ax.set_ylim(-0.5, len(ylabels) - 0.5)
ax.set(xticks=np.arange(len(xlabels)), yticks=np.arange(len(ylabels)),
       xticklabels=xlabels, yticklabels=ylabels)

ax.set_xticks(np.arange(len(xlabels)) - 0.5, minor=True)
ax.set_yticks(np.arange(len(ylabels)) - 0.5, minor=True)

ax.grid(which='minor')
ax.set_aspect("equal", "box")

R = [0.3] * len(s)
circles = [plt.Circle((xlabels_num_list[i], ylabels_num_list[i]), radius=r) for i, r in enumerate(R)]

norm = TwoSlopeNorm(vmin=0, vmax=1.0, vcenter=0.05)  # ← 改这里
col = PatchCollection(circles, array=c, cmap='Oranges_r', norm=norm, edgecolor='white', linewidth=1)

ax.add_collection(col)

ax.tick_params(axis='x', labelrotation=90)
ax.tick_params(axis='both', labelsize=6)

sns.despine(left=False, bottom=False, top=True, right=True)

cbar = fig.colorbar(col, shrink=0.1, aspect=10, orientation='horizontal', ticks=[0, 0.05, 0.25, 0.5, 1.0])  # ← 改这里
cbar.ax.set_xticklabels([0, 0.05, 0.25, 0.5, 1.0])  # ← 改这里

plt.tight_layout()
plt.show()

```

```python
import pandas as pd

# Load enrichment data
df_go = pd.read_csv("./analysis/gigtransformer-rownorm/t2ds_norm_enrichment_pathway.csv")
df_go = df_go.rename(columns={"P-Value": "PValue", "User Ids": "Genes"})
df_go["Genes"] = df_go["Genes"].str.replace(",", ", ")

# Keep only GOTERM_BP_DIRECT
go_bp = df_go[df_go["Category"] == "GOTERM_BP_DIRECT"]

# Must-include terms
must_include = [
    "positive regulation of canonical NF-kappaB signal transduction",
    "pyroptotic inflammatory response",
    "apoptotic process"
]

# Filter: FDR < 0.001, top 27 by PValue
go_top = go_bp[go_bp["FDR"] < 0.001].nsmallest(27, "PValue")

# Force include must-have terms
must_rows = go_bp[go_bp["Term"].isin(must_include)]
go_filtered = pd.concat([go_top, must_rows]).drop_duplicates(subset="Term").reset_index(drop=True)

print(f"GO BP filtered: {len(go_filtered)} terms")
go_filtered[["Term", "PValue", "FDR"]]
```

```python
# Expand Genes column into one row per gene
go_filtered["Genes"] = go_filtered["Genes"].str.split(", ")

rows = []
for _, row in go_filtered.iterrows():
    for gene in row["Genes"]:
        rows.append({"Term": row["Term"], "Gene": gene})

links_df_go = pd.DataFrame(rows)
display(links_df_go)
```

```python
from pycirclize import Circos
from pycirclize.parser import Matrix

# Build from-to table
fromto_go = links_df_go[["Term", "Gene"]].copy()
fromto_go = fromto_go.rename(columns={"Term": "from", "Gene": "to"})
fromto_go["value"] = [1] * len(fromto_go)

matrix_go = Matrix.parse_fromto_table(fromto_go)

circos_go = Circos.initialize_from_matrix(
    matrix_go,
    space=1,
    label_kws=dict(size=10),
)

fig = circos_go.plotfig()
```

```python
import plotly.graph_objects as go
import pandas as pd

# 给 GO term 分功能大类（Type）
go_category_mapping = {
    "Inflammation & Immune": [
        "pyroptotic inflammatory response",
        "positive regulation of canonical NF-kappaB signal transduction",
        "inflammatory response",
        "positive regulation of inflammatory response",
        "positive regulation of interleukin-1 beta production",
        "defense response to virus",
        "defense response to bacterium",
        "innate immune response",
    ],
    "Cell Death": [
        "apoptotic process",
        "positive regulation of apoptotic process",
        "regulation of apoptotic process",
        "programmed cell death",
        "necroptotic process",
        "execution phase of apoptosis",
    ],
    "Signaling": [
        "signal transduction",
        "positive regulation of I-kappaB kinase/NF-kappaB signaling",
        "positive regulation of ERK1 and ERK2 cascade",
        "MAPK cascade",
        "Wnt signaling pathway",
        "Hippo signaling",
    ],
}

def assign_go_category(term):
    for category, terms in go_category_mapping.items():
        for t in terms:
            if t.lower() in term.lower():
                return category
    return "Other"

links_df_go["Type"] = links_df_go["Term"].apply(assign_go_category)
links_df_go["Value"] = 1

# Build label list: Type → Term → Gene
labels = (
    links_df_go["Type"].unique().tolist()
    + links_df_go["Term"].unique().tolist()
    + links_df_go["Gene"].unique().tolist()
)
label_to_index = {label: idx for idx, label in enumerate(labels)}

links_df_go["Type_Index"]    = links_df_go["Type"].map(label_to_index)
links_df_go["Pathway_Index"] = links_df_go["Term"].map(label_to_index)
links_df_go["Gene_Index"]    = links_df_go["Gene"].map(label_to_index)

# Colors per Type
type_palette = {
    "Inflammation & Immune": "rgba(214, 39, 40, 0.5)",
    "Cell Death":            "rgba(31, 119, 180, 0.5)",
    "Signaling":             "rgba(44, 160, 44, 0.5)",
    "Other":                 "rgba(148, 103, 189, 0.5)",
}

link_colors = links_df_go["Type"].map(type_palette).tolist()

links = pd.concat([
    pd.DataFrame({
        "source": links_df_go["Type_Index"],
        "target": links_df_go["Pathway_Index"],
        "value":  links_df_go["Value"],
        "color":  link_colors,
    }),
    pd.DataFrame({
        "source": links_df_go["Pathway_Index"],
        "target": links_df_go["Gene_Index"],
        "value":  links_df_go["Value"],
        "color":  link_colors,
    }),
])

fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=5,
        thickness=20,
        line=dict(color="black", width=0.5),
        label=labels,
    ),
    link=dict(
        source=links["source"],
        target=links["target"],
        value=links["value"],
        color=links["color"],
    )
)])

fig.update_layout(
    title_text="GO BP Pathway Enrichment - Sankey Diagram",
    font_size=10,
    height=900,
)
fig.show()
```
